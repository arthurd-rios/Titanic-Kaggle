```python
tab = pd.crosstab(df['Pclass'], df['Survived'])

print(tab)
```

```python
for pclass, line in tab.iterrows():

    not_survived = line[0]
    survived = line[1]

    print(f"Survival rate for class {pclass}: {100*survived/(survived+not_survived):.2f}%")

# The higher the class, the higher the survival rate
```

```python
tab = pd.crosstab(df['Sex'], df['Survived'])

print(tab)
```

```python
for Sex, line in tab.iterrows():

    not_survived = line[0]
    survived = line[1]

    print(f"Survival rate for {Sex}: {100*survived/(survived+not_survived):.2f}%")

# Female has higher survival rate (they tried to save more womans and girls)
```

```python
conditions = [
    (df['Age'] < 12),
    (df['Age'] >= 12) & (df['Age'] < 18),
    (df['Age'] >= 18) & (df['Age'] < 60),
    (df['Age'] >= 60)
]

group = ['kid', 'teen', 'adult', 'elderly']

df['AgeGroup'] = np.select(conditions, group, default='unknown')
```

```python
tab = pd.crosstab(df['AgeGroup'], df['Survived'])

print(tab)
```

```python
for AgeGroup, line in tab.iterrows():

    not_survived = line[0]
    survived = line[1]

    print(f"Survival rate for age group {AgeGroup}: {100*survived/(survived+not_survived):.2f}%")

# The younger the age group, the higher the survival rate (they tried to save more kids and teens)
```

```python
df['NameTitle'] = df['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
```

```python
mask = df['AgeGroup'] == 'unknown'

df.loc[mask, 'NameTitle'].value_counts()

# Mr = Adult
# Miss = Woman
# Mrs = Adult
# Master = Kid
# Dr = Probably adult, but maybe Elderly
```

```python
mask = df['Embarked'].isna()

df[mask]
```

```python
print(df.groupby('Embarked')['Fare'].describe())

# For a 80.0 Fare, the Embarked value is probably C
```

```python
df['Embarked'] = df['Embarked'].fillna('C')
```

```python
TitleAge = {"Mr": "adult", "Mrs": "adult", "Dr": "adult", "Master": "kid"}

for title, age in TitleAge.items():

    mask = (df['AgeGroup'] == "unknown") & (df['NameTitle'] == title)

    df.loc[mask, 'AgeGroup'] = age
```

```python
df['TicketCount'] = df.groupby('Ticket').transform('size')
df['IndividualFare'] = df['Fare']/df['TicketCount']

print(df.groupby('Pclass')['IndividualFare'].median())
```

```python
# Mapping "Miss" to AgeGroup categories

mask = (df['AgeGroup'] == "unknown") & (df['NameTitle'] == "Miss")

df[mask]
```

```python
mask = (df['AgeGroup'] == "unknown") & (df['NameTitle'] == "Miss") & (df['TicketCount'] == 1)

df.loc[mask, 'AgeGroup'] = 'adult'

# Miss traveling alone is assumed to be an adult
```

```python
mask = (df['AgeGroup'] == "unknown") & (df['NameTitle'] == "Miss") & (df['Parch'] == 0)

df.loc[mask, 'AgeGroup'] = 'adult'
```

```python
mask = (df['AgeGroup'] == "unknown") & (df['NameTitle'] == "Miss")

df.loc[mask, 'AgeGroup'] = 'kid'
```

```python
PclassDeckLevel = {1: 3, 2: 2, 3: 1}

df['DeckLevel'] = df['Pclass'].map(PclassDeckLevel)
```
